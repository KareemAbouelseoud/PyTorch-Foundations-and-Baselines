### Notes

### .tensor() creates a copy of the data

```python
array = np.array([1, 2, 3])
tensor = torch.tensor(array)
```

```python
tensor[0] = 999
print("Numpy array:", array)
print("Torch tensor:", tensor)
```

#### To avoid that, we can use torch.as_tensor() to convert np array to tensor

```python
array = np.array([1, 2, 3])
tensor = torch.as_tensor(array)
```

```python
tensor[0] = 999
print("Numpy array:", array)
print("Torch tensor:", tensor)
```

## Random Tensors

### Rand()

##### rand() gives a uniform distribution [0,1)
##### this can be used for Probabilities, dropout mask, or exploration noise

```python
random = torch.rand(3)
random
```

- ##### size of the tensor is always needed
- ##### by default, the type is float32

```python
random.dtype
```

```python
torch.rand(size=(3, 4), dtype=torch.float16)
```

### Randn()

#### randn gives a normal distribution (mean of 0 and std of 1) (-inf,+inf)
#### This is what we will mostly use for RL, and DL in general

```python
random = torch.randn([3,255,255])
random
```

### Randint()

#### Discrete Uniform Integers [low,high)
#### Sampling random batch indices, discrete action spaces

```python
# High and low parameters are required
torch.randint(size=(3, 4),high=10, low=0)
```

### Random state in pytorch

```python
# Global seed
torch.manual_seed(42)
W=torch.rand(3, 4)
W
```

```python
g =  torch.Generator().manual_seed(43)
W = torch.rand(3, 4, generator=g)
W
```

```python
W.ndim,W.shape
```

## Devices

```python
print(torch.cuda.is_available())        # Returns True if CUDA GPU is ready
print(torch.cuda.device_count())        # Number of available GPUs
print(torch.cuda.get_device_name(0))
```

```python
device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"
device
```

### Default initilizations use CPU, to transfer to GPU use this

```python
# BAD: Allocates on CPU, then copies memory to GPU
x = torch.rand(3, 4)
# Not in place, you need to assign it back to x or another variable (only for tensors, not for modules)
x = x.to(device)
x
```

```python
# GOOD: Allocates directly in GPU memory
x = torch.rand(3, 4, device=device)
x
```

### Operations Require Same Device Tensor

```python
a = torch.ones(2, 2, device="cpu")
b = torch.ones(2, 2, device="cuda")

# CRASH! Raises RuntimeError
# c = a + b 

# FIX: Bring 'a' to 'cuda' first
c = a.to("cuda") + b
```

### Back To Numpy

```python
x = torch.randn(3, 3, device="cuda")

# BAD: Raises RuntimeError (GPU memory + autograd attached)
# np_arr = x.numpy()

# GOOD: Detaches graph, copies to CPU memory, then converts to NumPy
np_arr = x.cpu().numpy()
np_arr
```

## Like functions

#### There are like functions that create a tensor with all the attributes such as device or shape of the passed tensor

```python
state = torch.tensor([1.5, 2.0], device="cuda", dtype=torch.float32)
# Not the best way and prone to errors if you forget the device or dtype
noise = torch.rand(state.shape, device=state.device, dtype=state.dtype)
# Better Approach
noise = torch.rand_like(state)
```
